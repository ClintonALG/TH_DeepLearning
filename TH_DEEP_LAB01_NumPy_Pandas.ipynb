```python
#bai_1

# Tạo mảng 3x3 với giá trị từ 0 đến 20
arr1 = np.random.randint(0, 21, size=(3, 3))
print("Mảng 1:\n", arr1)

# Kiểm tra tất cả các phần tử có khác 0 hay không
is_all_nonzero = np.all(arr1 != 0)
print("Tất cả phần tử khác 0:", is_all_nonzero)
```

```python
#bai3

# Mảng 1 chiều
x1 = np.array([1, 5, 10])
y1 = np.array([2, 5, 8])
print("So sánh 1 chiều (x1 > y1):", np.greater(x1, y1))
print("So sánh 1 chiều (x1 <= y1):", np.less_equal(x1, y1))

# Mảng 2 chiều
x2 = np.array([[1, 2], [3, 4]])
y2 = np.array([[1, 3], [2, 4]])
print("So sánh 2 chiều (x2 >= y2):\n", np.greater_equal(x2, y2))
print("So sánh 2 chiều (x2 < y2):\n", np.less(x2, y2))
```
