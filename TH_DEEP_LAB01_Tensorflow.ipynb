```python
#bai10

x10 = tf.random.uniform([5], minval=1, maxval=10)
y10 = tf.random.uniform([5], minval=1, maxval=10)

print("x =", x10.numpy())
print("y =", y10.numpy())

# Sử dụng tf.where để kiểm tra điều kiện x < y
t10 = tf.where(x10 < y10, x10 + y10, x10 - y10)
print("Bài 10:\n", t10.numpy())
```

```python
#bai11

x11 = tf.random.uniform([3, 2], minval=5.0, maxval=10.0)
y11 = tf.random.uniform([2, 3], minval=15.0, maxval=20.0)

# Giữ lại đường chéo của x (xóa các phần tử không thuộc đường chéo chính)
x_diag = tf.linalg.band_part(x11, 0, 0)

# Chuyển vị y thành (3, 2) để có thể cộng với x
y_transpose = tf.transpose(y11)

t11 = x_diag + y_transpose
print("Bài 11:\n", t11.numpy())
```

```python
#bai13

class SumModel(tf.Module):
    @tf.function
    def __call__(self, x, y):
        return x + y

model_13 = SumModel()
print("Bài 13:", model_13(tf.constant(5), tf.constant(10)).numpy())
```

```python
#bai14

class DiffProdModel(tf.Module):
    @tf.function
    def __call__(self, x, y):
        return x - y, x * y

model_14 = DiffProdModel()
diff, prod = model_14(tf.constant(8), tf.constant(3))
print("Bài 14 - Hiệu:", diff.numpy(), "Tích:", prod.numpy())
```

```python
#bai15

class RectanglePerimeter(tf.Module):
    @tf.function
    def __call__(self, a, b):
        return (a + b) * 2

model_15 = RectanglePerimeter()
print("Bài 15 (a=3, b=4):", model_15(tf.constant(3), tf.constant(4)).numpy())
```

```python
#bai16

class BoxSurfaceArea(tf.Module):
    @tf.function
    def __call__(self, a, b, c):
        s_xq = (a + b) * c * 2
        s_tp = (a * b) * 2 + s_xq
        return s_tp

model_16 = BoxSurfaceArea()
print("Bài 16 (a=2, b=3, c=4):", model_16(tf.constant(2), tf.constant(3), tf.constant(4)).numpy())
```

```python
#bai17

class TrapezoidParallelProof(tf.Module):
    @tf.function
    def __call__(self, AD, BC, AB, CD):
        cond1 = tf.equal(AD, BC)
        cond2 = tf.equal(AB, CD)
        # Trả về True nếu cả 2 điều kiện đều đúng
        return tf.logical_and(cond1, cond2)

model_17 = TrapezoidParallelProof()
print("Bài 17:", model_17(tf.constant(5), tf.constant(5), tf.constant(10), tf.constant(10)).numpy())
```

```python
#bai18

class TrapezoidBaseProof(tf.Module):
    @tf.function
    def __call__(self, AD, BC, is_parallel):
        cond1 = tf.equal(AD, BC)
        return tf.logical_and(cond1, is_parallel)

model_18 = TrapezoidBaseProof()
print("Bài 18:", model_18(tf.constant(4), tf.constant(4), tf.constant(True)).numpy())
```

```python
#bai19

class TriangleMidsegmentProof(tf.Module):
    @tf.function
    def __call__(self, AM, MB, MN, BC):
        # Đường trung bình cần thỏa mãn MN = BC/2
        cond1 = tf.equal(AM, MB)
        cond2 = tf.equal(MN, BC / 2.0)

        return tf.logical_and(cond1, cond2)

model_19 = TriangleMidsegmentProof()
print("Bài 19:", model_19(tf.constant(3.0), tf.constant(3.0), tf.constant(5.0), tf.constant(10.0)).numpy())
```
